# Convertion of FEMTO dataset to CMAPSS-like dataset

To transfer CMAPSS model to the FEMTO dataset, a simple convertion is applied on the original FEMTO datasets with the following step : each segment is compressed into 16 fetatures each, representating 8 statistical time-domain features and 8 statistical frequency-domain features using fast fourier transform. The statistical features are (for time and frequency) : mean, std, RMS, kurtosis, max of abs, min of abs, skewness, peak-to-peak.

```python
%reset -f
```

```python
import sys
sys.path.append('../')
import pyrul
import pandas as pd
import numpy as np
from pyrul.utils.compat import convert_femto_to_cmapss
```

```python
femto_train_data_path = "../datasets/femto/training_data.csv"
femto_test_data_path = "../datasets/femto/test_data.csv"
```

```python
df_train = pd.read_csv(femto_train_data_path, sep=",")
df_test = pd.read_csv(femto_test_data_path, sep=",")
```

```python
df_train.head()
```

```python
df_train_new = convert_femto_to_cmapss(df_train)
df_test_new = convert_femto_to_cmapss(df_test)
```
