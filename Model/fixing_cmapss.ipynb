# Fixing C-MAPSS dataset

This notebook applies a basic preprocessing of the CMAPSS FD001 dataset to add column names. The saved datasets are called "train_FD001_fixed.txt" and "test_FD001_fixed.txt".

```python
%reset -f
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## 1. General Data Processing

```python
# Importing and fixing raw data of subdataset FD001
df_train = pd.read_csv('../cmapss/train_FD001.txt', sep=" ", header=None)
df_test = pd.read_csv('../cmapss/test_FD001.txt', sep=" ", header=None)

# Removing last 2 columns
df_train.drop(columns=[26,27], inplace=True)
df_test.drop(columns=[26,27], inplace=True)
```

Data Set: FD001

Train trajectories: 100

Test trajectories: 100

Conditions: ONE (Sea Level)

Fault Modes: ONE (High-pressure Core (HPC) Degradation)

```python
# Adding train and test RUL (assuming unit start from 1 to 100)
df_train['RUL'] = df_train.groupby('unit')['cycle'].apply(lambda x: x.iloc[::-1]).reset_index(drop=True) - 1
df_test['RUL'] = df_test.groupby('unit')['cycle'].apply(lambda x: x.iloc[::-1]).reset_index(drop=True) - 1

last_ruls = np.loadtxt('../cmapss/RUL_FD001.txt')
for e in df_test["unit"].unique():
    row_idx = df_test[df_test['unit']==e].index
    df_test.loc[row_idx, "RUL"] += last_ruls[e-1]
```

```python
# Saving
df_train.to_csv('./train_FD001_fixed.txt', sep=" ")
df_test.to_csv('./cmapss/test_FD001_fixed.txt', sep=" ")
```
